```python
import os
import csv
import json

# Define the paths for the CSV and JSON folders
csv_folder = os.path()
json_folder = 'Dados Taxboard - Paula Torres/jsons'

# Ensure the output directory exists
os.makedirs(json_folder, exist_ok=True)

# Get the list of CSV files in the CSV folder
csv_files = [file for file in os.listdir(csv_folder) if file.endswith('.csv')]

# Iterate over each CSV file
for csv_file in csv_files:
    try:
        # Create the path for the CSV file
        csv_path = os.path.join(csv_folder, csv_file)
        
        # Create the path for the JSON file
        json_file = os.path.splitext(csv_file)[0] + '.json'
        json_path = os.path.join(json_folder, json_file)

        # Try reading the CSV with utf-8 encoding first
        try:
            with open(csv_path, 'r', encoding='utf-8') as file:
                csv_data = csv.DictReader(file)
                data = [row for row in csv_data]

        # If utf-8 fails, fallback to latin-1
        except UnicodeDecodeError:
            with open(csv_path, 'r', encoding='latin-1') as file:
                csv_data = csv.DictReader(file)
                data = [row for row in csv_data]

        # Write the JSON data to the file
        with open(json_path, 'w', encoding='utf-8') as json_file:
            json.dump(data, json_file, indent=4)
        
        print(f"Converted {csv_file} to {json_file}")
    
    except Exception as e:
        print(f"Failed to convert {csv_file}: {e}")
```
