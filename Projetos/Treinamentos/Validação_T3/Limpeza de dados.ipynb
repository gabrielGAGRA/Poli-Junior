# Analisando customers.csv

```python
df_customers = pd.read_csv('customers.csv')
print(df_customers.info())
```

Daqui vemos que existem informações nulas dependendo da coluna


```python
df_customers_clean = df_customers.drop_duplicates()
print(df_customers_clean.info())
```

Não existem linhas duplicadas

```python
filtro = df_customers_clean[df_customers_clean['customer_city'].isnull()]
filtro
```

```python
filtro = df_customers_clean[df_customers_clean['customer_state'].isnull()]
filtro
```

```python
# Preenchendo os NaNs em 'customer_state' com base nas informações de outras linhas que possuem o mesmo 'customer_city'
df_customers_clean['customer_state'] = df_customers_clean.groupby('customer_city')['customer_state'].transform(lambda x: x.ffill().bfill())

# Exibir o DataFrame após preenchimento
print(df_customers_clean.info())

df_customers_clean.to_csv('customers.csv', index=False)
```

Tratamento rápido dos nan da coluna costume_state, pegando cidades já presentes que estão relacionadas a estados e preenchendo a partir disso
