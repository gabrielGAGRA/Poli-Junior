```python
df = pd.read_csv('base_atualizada.csv')
display(df)
```

```python
import pandas as pd

# Convertendo as colunas para o formato datetime
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])
df['order_approved_at'] = pd.to_datetime(df['order_approved_at'])
df['order_delivered_carrier_date'] = pd.to_datetime(df['order_delivered_carrier_date'])
df['order_delivered_customer_date'] = pd.to_datetime(df['order_delivered_customer_date'])

# Criando as colunas com a diferença de tempo em dias
df['purchase_to_approval_minutes'] = (df['order_approved_at'] - df['order_purchase_timestamp']).dt.total_seconds() / 60
df['approval_to_carrier_minutes'] = (df['order_delivered_carrier_date'] - df['order_approved_at']).dt.total_seconds() / 60
df['carrier_to_customer_minutes'] = (df['order_delivered_customer_date'] - df['order_delivered_carrier_date']).dt.total_seconds() / 60

# Exibindo as primeiras linhas do DataFrame
df.head()
```

```python
dfh = df[['price', 'freight_value', 'product_weight_g', 'delivery_time (days)', 
              'payment_value', 'product_volume', 'purchase_to_approval_days', 
              'approval_to_carrier_days', 'carrier_to_customer_days','distancia_km']]

# Exibindo as primeiras linhas do novo DataFrame
dfh.head()
```

```python
dfh.info()
```

```python
# Removendo linhas com valores nulos na coluna 'delivery_time'
dfh = dfh.dropna(subset=['delivery_time (days)'])

# Exibindo as primeiras linhas do DataFrame limpo
dfh.info()
```

```python
# Simulando novamente o DataFrame para a execução do código de limpeza de outliers
import pandas as pd
import numpy as np

# Removendo outliers das colunas especificadas usando o método IQR
def remove_outliers_iqr(df, columns):
    for column in columns:
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        df = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    return df

# Lista das colunas para remover os outliers
columns_to_clean = ['freight_value', 'delivery_time (days)', 'approval_to_carrier_days', 
                    'carrier_to_customer_days']

# Aplicando a função de remoção de outliers
novo_df_cleaned = remove_outliers_iqr(novo_df, columns_to_clean)

# Exibindo as primeiras linhas do DataFrame limpo
novo_df_cleaned
```

```python
correlation_matrix = novo_df_cleaned.corr()

# Criando o heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Heatmap de Correlação do DataFrame")
plt.show()
```
