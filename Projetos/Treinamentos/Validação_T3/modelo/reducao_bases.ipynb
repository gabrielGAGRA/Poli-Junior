```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib as plt
```

```python
df_treino = pd.read_csv('FBase_Treino_Consolidada.csv')
```

```python
pd.set_option('display.max_columns', None)
display(df_treino)
```

```python
df_treino = df_treino.drop_duplicates(subset=['order_id', 'product_id'])
display(df_treino)
```

```python
df_treino = df_treino.sort_values(
    by=['order_id', 'distancia_km', 'purchase_to_approval_minutes', 'freight_value', 'product_volume'], 
    ascending=[False, False, False, False, False]
)

# Mantendo apenas a linha com os maiores valores por 'order_id'
df_treino = df_treino.groupby('order_id').first().reset_index()
display(df_treino)
```

```python
# Número de order_ids diferentes
num_order_ids = df_treino['order_id'].nunique()

print("Número de order_ids diferentes:", num_order_ids)
```

```python
df_teste = pd.read_csv('FBase_Teste_Consolidada.csv')
```

```python
pd.set_option('display.max_columns', None)
display(df_teste)
```

```python
df_teste = df_teste.drop_duplicates(subset=['order_id', 'product_id'])
display(df_teste)
```

```python
df_teste = df_teste.sort_values(
    by=['order_id', 'distancia_km', 'purchase_to_approval_minutes', 'freight_value', 'product_volume'], 
    ascending=[False, False, False, False, False]
)

# Mantendo apenas a linha com os maiores valores por 'order_id'
df_teste = df_teste.groupby('order_id').first().reset_index()
display(df_teste)
```

```python
df_teste.rename(columns={'purchase_to_approval_minutes': 'purchase_to_approval_days'}, inplace=True)
df_treino.rename(columns={'purchase_to_approval_minutes': 'purchase_to_approval_days'}, inplace=True)
```

```python
# Removendo as colunas especificadas
df_treino = df_treino.drop(columns=[
    'customer_id',
    'order_status',
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'delivery_time_model (days)',
    'customer_zip_code_prefix',
    'order_item_id',
    'product_id',
    'seller_id',
    'shipping_limit_date',
    'seller_zip_code_prefix'
])

# Exibindo o DataFrame atualizado
display(df_treino)
```

```python
# Removendo as colunas especificadas
df_teste = df_teste.drop(columns=[
    'customer_id',
    'order_purchase_timestamp',
    'order_approved_at',
    'customer_zip_code_prefix',
    'order_item_id',
    'product_id',
    'seller_id',
    'shipping_limit_date',
    'seller_zip_code_prefix'
])

display(df_teste)
```
